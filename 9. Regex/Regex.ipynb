# Regular Expressions (Regex) in Python

This notebook is a comprehensive introduction to regular expressions in Python. We'll cover both basic and advanced topics with hands-on examples.

We will use the `re` module for all regex operations.

```python
import re
```

## 1. Basic Syntax

A regex is a pattern that specifies a set of strings. You can use regex to search, match, and manipulate text.

Common symbols:
- `.` matches any character except newline
- `^` matches the start of a string
- `$` matches the end of a string
- `*` matches 0 or more repetitions
- `+` matches 1 or more repetitions
- `?` matches 0 or 1 repetition
- `[]` specifies a character class
- `|` is used for OR
- `()` groups parts of the pattern

```python
# Match a word that starts with 'h' and ends with 'o'
pattern = r'h.*o'
text = 'hello world, halo, hero'
matches = re.findall(pattern, text)
print(matches)
```

## 2. Character Classes

- `[abc]` matches 'a', 'b', or 'c'
- `[a-z]` matches any lowercase letter
- `\d` matches any digit (same as `[0-9]`)
- `\D` matches any non-digit
- `\w` matches alphanumeric (letters + digits + `_`)
- `\W` matches non-word characters
- `\s` matches whitespace
- `\S` matches non-whitespace

```python
# Find all digits
pattern = r'\d+'
text = 'Phone: 1234, Code: 5678'
print(re.findall(pattern, text))
```

## 3. Anchors

Anchors don't match characters but positions:
- `^` matches the start of a string
- `$` matches the end of a string

```python
# Match only if 'hello' is at the start
pattern = r'^hello'
print(bool(re.match(pattern, 'hello world')))
```

## 4. Groups and Capturing

Use parentheses `()` to create groups and capture specific parts of a string.

```python
# Extract first and last name
pattern = r'(\w+) (\w+)'
text = 'John Smith'
match = re.match(pattern, text)
if match:
    print('First name:', match.group(1))
    print('Last name:', match.group(2))
```

## 5. Quantifiers

- `*` matches 0 or more
- `+` matches 1 or more
- `?` matches 0 or 1
- `{n}` matches exactly n times
- `{n,}` matches at least n times
- `{n,m}` matches between n and m times

```python
# Match a pattern that repeats 3 to 5 times
pattern = r'\d{3,5}'
text = '123 12345 1234567'
print(re.findall(pattern, text))
```

## 6. Substitution with `re.sub`

You can replace matched patterns using `re.sub()`.

```python
# Mask phone numbers
text = 'Call me at 123-456-7890 or 987-654-3210'
masked = re.sub(r'\d{3}-\d{3}-\d{4}', 'XXX-XXX-XXXX', text)
print(masked)
```

## 7. Lookaheads and Lookbehinds

These allow you to assert what comes before or after a pattern **without including it** in the match.

- `(?=...)` Positive lookahead
- `(?!...)` Negative lookahead
- `(?<=...)` Positive lookbehind
- `(?<!...)` Negative lookbehind

```python
# Match numbers only if they are followed by 'USD'
text = '100USD 200EUR 300USD'
pattern = r'\d+(?=USD)'
print(re.findall(pattern, text))
```

## 8. Summary
