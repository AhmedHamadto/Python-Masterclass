# Introduction to Pandas

This notebook introduces the basics and some advanced features of `pandas`, a powerful data analysis library in Python.

We'll cover data creation, selection, filtering, grouping, and manipulation.

```python
import pandas as pd
import numpy as np
```

## 1. Creating DataFrames

You can create a DataFrame from a dictionary, list of lists, or from a CSV file.

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'Salary': [50000, 60000, 70000, 80000]
}
df = pd.DataFrame(data)
df
```

## 2. Basic Operations

You can view, describe, and manipulate the data easily.

```python
print(df.head())  # First 5 rows
print(df.describe())  # Summary statistics
```

## 3. Indexing and Selection

Use `.loc[]` for label-based indexing and `.iloc[]` for position-based indexing.

```python
print(df.loc[1])  # Second row by label
print(df.iloc[2])  # Third row by index
```

## 4. Filtering Data

You can filter rows based on conditions.

```python
df[df['Age'] > 30]
```

## 5. Adding and Dropping Columns

```python
# Add a new column
df['Tax'] = df['Salary'] * 0.1

# Drop a column
df = df.drop('Tax', axis=1)
df
```

## 6. Grouping and Aggregation

Group data and perform aggregations like sum, mean, etc.

```python
df['Department'] = ['HR', 'IT', 'IT', 'HR']
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
```

## 7. Handling Missing Data

Use `isna()`, `fillna()`, and `dropna()` to manage missing values.

```python
df.loc[1, 'Salary'] = np.nan
print(df)

# Fill missing values
df['Salary'] = df['Salary'].fillna(0)
```

## 8. Merging and Joining

Combine DataFrames using `merge()` and `concat()`.

```python
df1 = pd.DataFrame({'ID': [1, 2], 'Score': [90, 95]})
df2 = pd.DataFrame({'ID': [1, 2], 'Grade': ['A', 'A+']})
merged = pd.merge(df1, df2, on='ID')
print(merged)
```

## 9. Summary

In this notebook, we covered key `pandas` features including:
- Creating and viewing data
- Indexing and filtering
- Grouping and aggregation
- Handling missing data
- Merging and joining DataFrames

Explore these functions with your own datasets to build confidence.
