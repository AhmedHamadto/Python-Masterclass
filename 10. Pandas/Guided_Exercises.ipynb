```python
import pandas as pd
import numpy as np
```

## Exercise 1: Create a DataFrame

Create a DataFrame with columns `Name`, `Age`, and `Score`.


```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [24, 30, 35, 28],
    'Score': [88, 92, 85, 90]
}
df = pd.DataFrame(data)
df
```

## Exercise 2: Filter Rows

Select rows where `Age` is greater than 25.


```python
df[df['Age'] > 25]
```

## Exercise 3: Add a New Column

Add a column called `Bonus` which is 10% of `Score`.


```python
df['Bonus'] = df['Score'] * 0.1
df
```

## Exercise 4: Group by Column

Assume you add a column `Department`. Find the average `Score` per department.


```python
df['Department'] = ['HR', 'IT', 'IT', 'HR']
df.groupby('Department')['Score'].mean()
```

## Exercise 5: Handle Missing Data

Introduce a missing value in the `Score` column and then fill it with the column mean.


```python
df.loc[1, 'Score'] = np.nan
mean_score = df['Score'].mean()
df['Score'] = df['Score'].fillna(mean_score)
df
```
