# Python Dictionaries

Dictionaries are unordered collections of key-value pairs. Each key must be unique, and they allow for fast access to data via keys.


## Creating a Dictionary

Use curly braces `{}` to define a dictionary with key-value pairs separated by colons.


```python
my_dict = {
    'name': 'Alice',
    'age': 25,
    'city': 'London'
}
print(my_dict)
```

## Accessing Values

You can access dictionary values using the key in square brackets or with the `.get()` method.


```python
print(my_dict['name'])        # Outputs: Alice
print(my_dict.get('age'))     # Outputs: 25
```

## Adding and Updating Items

Assign a value to a new key to add it. Reassign a value to update it.


```python
my_dict['age'] = 26           # Update age
my_dict['country'] = 'UK'     # Add new key-value pair
print(my_dict)
```

## Removing Items

Use `del`, `.pop()`, or `.popitem()` to remove items.


```python
del my_dict['city']           # Remove by key
print(my_dict)

removed_age = my_dict.pop('age')  # Removes and returns the value
print(f"Removed age: {removed_age}")
print(my_dict)
```

## Looping Through a Dictionary

Use `.items()`, `.keys()`, and `.values()` to loop through a dictionary.


```python
for key, value in my_dict.items():
    print(f"{key}: {value}")
```

## Useful Dictionary Methods

Explore some helpful methods available with dictionaries.


```python
print(my_dict.keys())     # All keys
print(my_dict.values())   # All values

# Update value
my_dict.update({'name': 'Bob'})
print(my_dict)
```

## Nesting Dictionaries

Dictionaries can contain other dictionaries as values.


```python
people = {
    'person1': {'name': 'Alice', 'age': 25},
    'person2': {'name': 'Bob', 'age': 30}
}

print(people['person1']['name'])  # Outputs: Alice
```

## Summary

- Dictionaries store key-value pairs
- Keys must be unique and immutable
- Use methods like `.get()`, `.update()`, `.pop()`, `.keys()`, `.values()`, `.items()`
- You can nest dictionaries for complex data structures
