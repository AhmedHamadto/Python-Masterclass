# Guided Exercises: Python Dictionaries

These exercises will help you practise using dictionaries in Python. Try to solve each problem yourself before running the code.


## Exercise 1: Create a Dictionary

Create a dictionary for a student that contains their name, student ID, and grade.


```python
student = {
    'name': 'Emma',
    'student_id': 'S12345',
    'grade': 'A'
}
print(student)
```

## Exercise 2: Access a Value

Print the student's grade using the correct key.


```python
print(student['grade'])
```

## Exercise 3: Add a New Key-Value Pair

Add a new key `email` to the student dictionary with the value `"emma@example.com"`.


```python
student['email'] = 'emma@example.com'
print(student)
```

## Exercise 4: Update a Value

Change the student's grade from `'A'` to `'A+'`.


```python
student['grade'] = 'A+'
print(student)
```

## Exercise 5: Remove a Key-Value Pair

Remove the `email` key from the dictionary.


```python
removed = student.pop('email')
print(f"Removed: {removed}")
print(student)
```

## Exercise 6: Loop Through Dictionary

Print all key-value pairs in the student dictionary.


```python
for key, value in student.items():
    print(f"{key}: {value}")
```

## Exercise 7: Create a Nested Dictionary

Create a dictionary of two students, each being a nested dictionary.


```python
students = {
    'student1': {'name': 'Emma', 'grade': 'A+'},
    'student2': {'name': 'Liam', 'grade': 'B'}
}
print(students['student2']['name'])  # Should print: Liam
```
